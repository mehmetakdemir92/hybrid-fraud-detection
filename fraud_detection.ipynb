Velocity Features: In general, when fraudsters get the card they do a lot transactions in such short time. Velocity features are very strong in fraud detection systems. Frauds tend to happen in nighttime, normal transactions seem like happening all day equally.

```python
df = df.sort_values("Time").reset_index(drop=True)
times = df["Time"].values

def count_last_window(times, window_seconds):
    start_idx = np.searchsorted(times, times - window_seconds, side="left")
    counts = np.arange(len(times)) - start_idx
    return counts

df["trans_last_1h"] = count_last_window(times, 3600)
df["trans_last_24h"] = count_last_window(times, 86400)
```

Ratio Features:
Fraudsters can make a transaction where the money is too much or too low.

```python
mean_amount = df["Amount"].mean()
df["amount_ratio"] = df["Amount"] / mean_amount
```

Zscore:

```python
df["amount_zscore"] = (df["Amount"] - df["Amount"].mean()) / df["Amount"].std()
```
